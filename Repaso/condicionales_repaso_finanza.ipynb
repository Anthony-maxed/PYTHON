```python
import pandas as pd
import numpy as np
import matplotlib as plt
import yfinance as yf
```

```python
#Extraemos data del ticker AMZN
#Puedes cambiar la fecha, el ticker para prácticar. Esto es con un solo ticker.
tickers = "AMZN"
data = yf.download(tickers, start= "2025-01-01", end = "2025-08-31")
data.head(5)
```

```python
#Solo nos quedamos con los precios de cierre
precios = data["Close"]
precios.head(5)
```

```python
#obtenemos la rentabilidades
returns = precios.pct_change().dropna()
returns 
```

```python
#obtenemos la rentabilidad anualizada
nday = returns.shape[0]
returns_anu = ((((1+returns).prod())**(252/nday))-1)*100
returns_anu
```

```python
#Primera manera de usar condicionales
re = returns_anu["AMZN"]
if re > 4:
    print("invertir")
if re < 4: 
    print("no invertir")
```

```python
#Segunda manera 
re = returns_anu["AMZN"]
if re > 4:
    print("invertir en su totalidad")
elif re > 2:
    print("invertir solo el 30")
else:
    print("no invertir")
```
